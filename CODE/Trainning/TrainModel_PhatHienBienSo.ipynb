```python
from ultralytics import YOLO
import os
import shutil

# 🧩 1️⃣ Tải model
model = YOLO("yolov8n.pt")

# 🧩 2️⃣ Train
results = model.train(
    data=f"{dataset.location}/data.yaml",
    epochs=50,          # số vòng lặp (tùy dataset)
    imgsz=640,          # kích thước ảnh
    batch=16,           # batch size
    name="ALPR_Dataset_v1",  # tên folder output
    device=0            # GPU
)

# 🧩 3️⃣ Xác định đường dẫn kết quả thực tế
save_dir = results.save_dir  # YOLO tự trả về thư mục lưu model
print(f"📁 YOLO lưu kết quả tại: {save_dir}")

# 🧩 4️⃣ Tự động sao lưu model sang Google Drive
drive_save_path = "/content/drive/MyDrive/ALPR_Dataset_v1"
os.makedirs(drive_save_path, exist_ok=True)

shutil.copytree(save_dir, drive_save_path, dirs_exist_ok=True)
print(f"✅ Đã sao lưu toàn bộ kết quả vào: {drive_save_path}")
```
